## 문제 1: 완전탐색(Brute-force) - 모음사전 (Level 2)

- **문제 출처**: [프로그래머스 모음사전](https://school.programmers.co.kr/learn/courses/30/lessons/84512)

---

### 문제 설명

사전에 알파벳 모음 'A', 'E', 'I', 'O', 'U'만을 사용하여 만들 수 있는, 길이 5 이하의 모든 단어가 수록되어 있습니다.
사전에서 첫 번째 단어는 "A"이고, 그다음은 "AA"이며, 마지막 단어는 "UUUUU"입니다.

단어 하나 `word`가 매개변수로 주어질 때, 이 단어가 사전에서 몇 번째 단어인지 return 하도록 solution 함수를 완성해주세요.

- `word`의 길이는 1 이상 5 이하입니다.
- `word`는 알파벳 대문자 'A', 'E', 'I', 'O', 'U'로만 이루어져 있습니다.

---

### 입출력 예시 및 작동 흐름 설명

| word | return |
| --- | --- |
| "AAAAE" | 6 |
| "AAAE" | 10 |
| "I" | 1563 |
| "EIO" | 1189 |

#### 진행 과정 및 상세 작동 흐름
1. 사전 순서 생성 흐름:
   - 1번째: "A"
   - 2번째: "AA"
   - 3번째: "AAA"
   - 4번째: "AAAA"
   - 5번째: "AAAAA"
   - 6번째: "AAAAE"
2. "AAAAE"의 경우 5번째 단어인 "AAAAA" 바로 다음이므로 6을 반환합니다.
3. "AAAE"의 경우:
   - "AAAA" 계열 6개("AAAA", "AAAAA", "AAAAE", "AAAAI", "AAAAO", "AAAAU") 탐색 종료 후 "AAAE"가 등장하므로 10번째가 됩니다.

````python
# [문제 설명]
# 사전에 알파벳 모음 'A', 'E', 'I', 'O', 'U'만을 사용하여 만들 수 있는, 길이 5 이하의 모든 단어가 수록되어 있습니다.
# 사전에서 첫 번째 단어는 "A"이고, 그다음은 "AA"이며, 마지막 단어는 "UUUUU"입니다.
# 단어 하나 word가 매개변수로 주어질 때, 이 단어가 사전에서 몇 번째 단어인지 return 하도록 solution 함수를 완성해주세요.

def solution(word):
    answer = 0
    weight = [781, 156, 31, 6, 1]
    value = {
        'A': 0, 'E': 1, 'I': 2, 'O': 3, 'U': 4
    }
    for i, v in enumerate(word):
        val = value[v]
        answer += weight[i] * val + 1
    return answer
````

## 문제 2: 깊이/너비 우선 탐색(DFS/BFS) - 무인도 여행 (Level 2)

- **문제 출처**: [프로그래머스 무인도 여행](https://school.programmers.co.kr/learn/courses/30/lessons/154540)

---

### 문제 설명

메리는 여름을 맞아 무인도로 여행을 가기 위해 지도를 보고 있습니다. 지도에는 바다와 무인도들에 대한 정보가 표시되어 있습니다.
지도는 1 x 1 크기의 사각형들로 이루어진 직사각형 격자 형태이며, 격자의 각 칸에는 `'X'` 또는 `1`에서 `9` 사이의 자연수가 적혀있습니다.

지도의 `'X'`는 바다를 나타내며, 숫자는 무인도를 나타냅니다.
이때, 상, 하, 좌, 우로 연결되는 땅들은 하나의 무인도를 이룹니다.
지도의 각 칸에 적힌 숫자는 식량을 나타내며, 상, 하, 좌, 우로 연결되는 칸에 적힌 숫자를 모두 합한 값은 해당 무인도에서 최대 며칠 동안 머물 수 있는지를 나타냅니다.

어떤 섬으로 놀러 갈지 못 정한 메리는 우선 각 섬에서 최대 며칠씩 머물 수 있는지 알아보고 놀러 갈 섬을 결정하려 합니다.

지도를 나타내는 문자열 배열 `maps`가 매개변수로 주어질 때, 각 섬에서 최대 며칠씩 머물 수 있는지 담은 배열을 오름차순으로 정렬하여 return 하는 solution 함수를 완성해주세요. 만약 지낼 수 있는 무인도가 없다면 `-1`을 배열에 담아 return 해주세요.

- 3 <= `maps`의 길이 <= 100
- 3 <= `maps[i]`의 길이 <= 100
- 지도의 모든 원소는 `'X'` 또는 1~9 사이의 자연수 문자입니다.

---

### 입출력 예시 및 작동 흐름 설명

| maps | return |
| --- | --- |
| ["X591X","X1X5X","X231X", "1XXX1"] | [1, 1, 5, 22] |
| ["XXX","XXX","XXX"] | [-1] |

#### 진행 과정 및 상세 작동 흐름 (첫 번째 예시 기준)
```text
X 5 9 1 X
X 1 X 5 X
X 2 3 1 X
1 X X X 1
```
1. 상하좌우로 인접하여 연결된 땅(연결 요소)들을 탐색합니다.
   - 섬 1: (0,1)=5, (0,2)=9, (0,3)=1, (1,1)=1, (2,1)=2, (2,2)=3, (2,3)=1 $ightarrow$ 총합 = 5 + 9 + 1 + 1 + 2 + 3 + 1 = **22**
   - 섬 2: (1,3)=5 $ightarrow$ 상하좌우가 모두 바다(X)이므로 독립된 섬 $ightarrow$ 총합 = **5**
   - 섬 3: (3,0)=1 $ightarrow$ 독립된 섬 $ightarrow$ 총합 = **1**
   - 섬 4: (3,4)=1 $ightarrow$ 독립된 섬 $ightarrow$ 총합 = **1**
2. 각 무인도에서 머물 수 있는 일수: `[22, 5, 1, 1]`
3. 이를 오름차순으로 정렬하면 **`[1, 1, 5, 22]`**가 됩니다.
4. 만약 지도의 모든 칸이 `'X'`라면 무인도가 없으므로 **`[-1]`**을 반환합니다.

````python
# [문제 설명]
# 지도를 나타내는 문자열 배열 maps가 주어집니다. ('X'는 바다, 1~9 숫자는 식량)
# 상, 하, 좌, 우로 연결된 숫자 칸들은 하나의 무인도를 이루며, 칸의 숫자 합이 머물 수 있는 날짜가 됩니다.
# 각 섬에서 최대 며칠씩 머물 수 있는지 담은 배열을 오름차순으로 정렬하여 return 하세요.
# 지낼 수 있는 무인도가 없다면 [-1]을 return 하세요.

def solution(maps):
    answer = []
    
    # 1. 상하좌우 이동을 위한 방향 벡터 정의
    # i=0: 위(-1, 0), i=1: 아래(1, 0), i=2: 왼쪽(0, -1), i=3: 오른쪽(0, 1)
    dx = [-1, 1, 0, 0]
    dy = [0, 0, -1, 1]
    
    # 2. 지도의 세로 크기(행 개수: r)와 가로 크기(열 개수: c) 구하기
    r = len(maps)        # 전체 줄(행) 수
    c = len(maps[0])     # 한 줄에 들어있는 글자(열) 수
    
    # 3. 방문 여부를 체크할 r x c 크기의 2차원 리스트 생성
    # 처음에는 어떤 칸도 방문하지 않았으므로 모든 칸을 False로 초기화합니다.
    visited = [[False] * c for _ in range(r)]

    # 4. 연결된 하나의 섬을 스택(Stack)으로 탐색하여 총 식량을 계산하는 DFS 함수
    # 재귀 함수 대신 스택을 사용하면 파이썬의 재귀 깊이 제한(런타임 에러)을 완벽히 피할 수 있습니다.
    def dfs(sx, sy):
        visited[sx][sy] = True         # 시작 칸 방문 처리
        total = int(maps[sx][sy])      # 시작 칸의 식량 값을 정수로 변환하여 누적 시작
        stack = [(sx, sy)]             # 앞으로 탐색할 땅의 좌표들을 담을 스택
        
        while stack:
            cx, cy = stack.pop()       # 스택에서 탐색할 현재 칸 (cx, cy) 꺼내기
            
            # 현재 칸을 기준으로 상, 하, 좌, 우 4방향 확인
            for i in range(4):
                nx = cx + dx[i]        # 이동할 다음 x 좌표
                ny = cy + dy[i]        # 이동할 다음 y 좌표
                
                # [유효성 검사 1] 다음 좌표가 지도 바깥으로 벗어나지 않는지 확인
                if 0 <= nx < r and 0 <= ny < c:
                    # [유효성 검사 2] 바다('X')가 아니고, 아직 가본 적 없는 땅인지 확인
                    if maps[nx][ny] != "X" and not visited[nx][ny]:
                        visited[nx][ny] = True           # 중복 방문 방지를 위해 즉시 방문 처리
                        total += int(maps[nx][ny])        # 식량을 총합에 더하기
                        stack.append((nx, ny))           # 다음 인접 땅 탐색을 위해 스택에 추가
                        
        return total                   # 이 섬에 연결된 모든 식량의 합 반환

    # 5. 지도의 모든 칸 (0,0) ~ (r-1, c-1)을 이중 for문으로 하나씩 스캔
    for i in range(r):
        for j in range(c):
            # 바다가 아니고, 이전에 다른 섬 탐색으로 들르지 않은 새로운 땅을 발견했을 때
            if maps[i][j] != "X" and not visited[i][j]:
                food = dfs(i, j)       # 해당 땅과 연결된 섬 전체의 식량을 합산
                answer.append(food)    # 결과 리스트에 섬의 식량 합 추가
                
    # 6. 결과 반환: 머물 수 있는 무인도가 있으면 오름차순 정렬, 하나도 없으면 [-1] 반환
    return sorted(answer) if answer else [-1]
````
