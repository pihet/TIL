세로길이가 `n` 가로길이가 `m`인 격자 모양의 땅 속에 석유가 묻혀 있습니다.
석유가 묻힌 땅은 `1`, 빈 땅은 `0`으로 표현됩니다.
상, 하, 좌, 우로 연결된 석유는 하나의 석유 덩어리를 이룹니다.

시추관을 수직으로 단 하나만 뚫을 수 있을 때, 시추관이 지나가는 모든 석유 덩어리의 석유량을 합산하여 뽑을 수 있습니다.

- 1 <= `land`의 세로 길이 `n` <= 500
- 1 <= `land`의 가로 길이 `m` <= 500
- `land[i][j]`는 0 또는 1입니다.

## 문제 2: 깊이/너비 우선 탐색(DFS/BFS) - 미로 탈출 (Level 2)

- **문제 출처**: [프로그래머스 미로 탈출](https://school.programmers.co.kr/learn/courses/30/lessons/159993)

---

### 문제 설명

1 x 1 크기의 칸들로 이루어진 직사각형 격자 형태의 미로에서 탈출하려고 합니다.
각 칸은 다음과 같이 표시됩니다.
- `'S'` : 시작 지점
- `'E'` : 출구
- `'L'` : 레버
- `'O'` : 통로 (이동 가능)
- `'X'` : 벽 (이동 불가)

시작 지점에서 출발하여 먼저 레버(`'L'`)가 있는 칸으로 이동하여 레버를 당긴 후, 출구(`'E'`)가 있는 칸으로 이동하여 탈출해야 합니다.
아직 레버를 당기지 않았더라도 출구가 있는 칸을 지나갈 수는 있습니다.
한 번 지나간 칸도 다시 방문할 수 있으며, 상, 하, 좌, 우로 1칸 이동하는 데 1초가 걸립니다.

미로를 나타내는 문자열 배열 `maps`가 주어질 때, 미로를 탈출하는 데 필요한 최소 시간을 return 하도록 solution 함수를 완성해주세요.
만약 탈출할 수 없다면 `-1`을 return 해주세요.

- 5 <= `maps`의 길이 <= 100
- 5 <= `maps[i]`의 길이 <= 100

---

### 입출력 예시

| maps | return |
| --- | --- |
| ["SOOOL","XXXXO","OOOOO","OXXXX","OOOOE"] | 16 |
| ["LOOXS","OOOOX","OOOOO","OOOOO","EOOOO"] | -1 |

```python
# [문제 설명]
# 직사각형 미로 maps가 주어집니다. ('S': 시작, 'E': 출구, 'L': 레버, 'O': 통로, 'X': 벽)
# 시작점(S)에서 출발하여 레버(L)를 당긴 뒤, 출구(E)로 탈출하는 최단 시간을 구하세요.
# 상, 하, 좌, 우로 1칸 이동 시 1초가 걸립니다.
# 탈출할 수 없다면 -1을 return 하세요.

from collections import deque


def solution(maps):
    # 격자 세로(n) 가로(m) 크기
    n = len(maps)
    m = len(maps[0])

    # 주요 지점 좌표 수집 (S: 시작, L: 레버, E: 출구)
    start = lever = end = None
    for i in range(n):
        for j in range(m):
            if maps[i][j] == 'S':
                start = (i, j)
            elif maps[i][j] == 'L':
                lever = (i, j)
            elif maps[i][j] == 'E':
                end = (i, j)

    # 상, 하, 좌, 우 이동 오프셋
    dx = [-1, 1, 0, 0]
    dy = [0, 0, -1, 1]

    # src -> dst 최단 거리를 BFS로 계산 (도달 불가 시 -1 반환)
    def bfs(src, dst):
        visited = [[False] * m for _ in range(n)]
        sx, sy = src
        visited[sx][sy] = True
        queue = deque([(sx, sy, 0)])  # (행, 열, 누적 이동 시간)

        while queue:
            x, y, dist = queue.popleft()
            if (x, y) == dst:  # 목표 지점 도착 시 그때의 거리가 최단 시간
                return dist

            for k in range(4):
                nx = x + dx[k]
                ny = y + dy[k]
                # 격자 범위 안 + 미방문 + 벽('X')이 아닌 칸만 이동
                if 0 <= nx < n and 0 <= ny < m and not visited[nx][ny] and maps[nx][ny] != 'X':
                    visited[nx][ny] = True
                    queue.append((nx, ny, dist + 1))

        return -1  # 큐를 모두 소진해도 dst에 못 닿으면 도달 불가

    # 1단계: 시작 -> 레버, 도달 불가면 전체 탈출 불가
    to_lever = bfs(start, lever)
    if to_lever == -1:
        return -1

    # 2단계: 레버 -> 출구, 도달 불가면 전체 탈출 불가
    to_exit = bfs(lever, end)
    if to_exit == -1:
        return -1

    # 두 구간 최단 시간의 합이 최소 탈출 시간
    return to_lever + to_exit
```
