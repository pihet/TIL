# 15일차 SQL 데이터 분석 학습

## 문제 1: 자동차 대여 기록에서 장기/단기 대여 구분하기 (Level 3)

- **문제 출처**: [프로그래머스 자동차 대여 기록에서 장기/단기 대여 구분하기](https://school.programmers.co.kr/learn/courses/30/lessons/151138)

---

### 문제 설명

`CAR_RENTAL_COMPANY_RENTAL_HISTORY` 테이블은 자동차 대여 기록 정보를 담은 테이블입니다.
`HISTORY_ID`, `CAR_ID`, `START_DATE`, `END_DATE`는 각각 대여 기록 ID, 자동차 ID, 대여 시작일, 대여 종료일을 의미합니다.

`CAR_RENTAL_COMPANY_RENTAL_HISTORY` 테이블에서 대여 시작일이 2022년 9월에 속하는 대여 기록에 대해서 대여 기간이 30일 이상이면 '장기 대여', 그렇지 않으면 '단기 대여'로 표시하는 컬럼(컬럼명: `RENT_TYPE`)을 추가하여 대여 기록 ID, 자동차 ID, 대여 시작일, 대여 종료일, 대여 구분을 출력하는 SQL문을 작성해주세요.
결과는 대여 기록 ID를 기준으로 내림차순 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### CAR_RENTAL_COMPANY_RENTAL_HISTORY (대여 기록)
| HISTORY_ID | CAR_ID | START_DATE | END_DATE |
| --- | --- | --- | --- |
| 1 | 4 | 2022-09-27 | 2022-10-26 |
| 2 | 3 | 2022-10-03 | 2022-11-04 |
| 3 | 2 | 2022-09-05 | 2022-09-05 |
| 4 | 1 | 2022-09-01 | 2022-09-30 |
| 5 | 3 | 2022-09-16 | 2022-10-13 |

#### 기대하는 출력 결과 (RESULT)
| HISTORY_ID | CAR_ID | START_DATE | END_DATE | RENT_TYPE |
| --- | --- | --- | --- | --- |
| 5 | 3 | 2022-09-16 | 2022-10-13 | 단기 대여 |
| 4 | 1 | 2022-09-01 | 2022-09-30 | 장기 대여 |
| 3 | 2 | 2022-09-05 | 2022-09-05 | 단기 대여 |
| 1 | 4 | 2022-09-27 | 2022-10-26 | 장기 대여 |


```sql
-- [문제 설명]
-- CAR_RENTAL_COMPANY_RENTAL_HISTORY 테이블에서 대여 시작일(START_DATE)이 2022년 9월인 기록만 조회합니다.
-- 대여 기간이 30일 이상이면 '장기 대여', 아니면 '단기 대여'로 표시하는 RENT_TYPE 컬럼을 추가합니다.
-- 출력 컬럼: HISTORY_ID, CAR_ID, START_DATE, END_DATE, RENT_TYPE
-- 결과는 HISTORY_ID 기준 내림차순 정렬합니다.
SELECT 
    HISTORY_ID,
    CAR_ID,
    DATE_FORMAT(START_DATE, '%Y-%m-%d') AS START_DATE,
    DATE_FORMAT(END_DATE, '%Y-%m-%d') AS END_DATE,
    CASE 
        WHEN DATEDIFF(END_DATE, START_DATE) + 1 >= 30 THEN '장기 대여'
        ELSE '단기 대여'
    END AS RENT_TYPE
FROM CAR_RENTAL_COMPANY_RENTAL_HISTORY
WHERE START_DATE LIKE '2022-09%'
ORDER BY HISTORY_ID DESC;
```

## 문제 2: 조건별로 분류하여 주문상태 출력하기 (Level 3)

- **문제 출처**: [프로그래머스 조건별로 분류하여 주문상태 출력하기](https://school.programmers.co.kr/learn/courses/30/lessons/131113)

---

### 문제 설명

`FOOD_ORDER` 테이블은 식품 주문 정보를 담은 테이블입니다.
`ORDER_ID`, `PRODUCT_ID`, `AMOUNT`, `PRODUCE_DATE`, `IN_DATE`, `OUT_DATE`, `FACTORY_ID`, `WAREHOUSE_ID`는 각각 주문 ID, 제품 ID, 수량, 생산일자, 입고일자, 출고일자, 공장 ID, 창고 ID를 의미합니다.

`FOOD_ORDER` 테이블에서 2022년 5월 1일을 기준으로 주문 ID, 제품 ID, 출고일자, 출고여부를 조회하는 SQL문을 작성해주세요.
출고여부(컬럼명: `출고여부`)는 다음과 같이 표시합니다.

- 출고일자가 2022년 5월 1일 이전이면(포함) '출고완료'로 표시
- 출고일자가 2022년 5월 1일 이후면 '출고대기'로 표시
- 출고일자가 없으면(NULL) '출고미정'으로 표시

결과는 주문 ID를 기준으로 오름차순 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### FOOD_ORDER (식품 주문)
| ORDER_ID | PRODUCT_ID | AMOUNT | PRODUCE_DATE | IN_DATE | OUT_DATE | FACTORY_ID | WAREHOUSE_ID |
| --- | --- | --- | --- | --- | --- | --- | --- |
| OD00000051 | P0002 | 4000 | 2022-04-01 | 2022-04-21 | 2022-04-21 | FT19970003 | WH0005 |
| OD00000052 | P0003 | 2500 | 2022-04-10 | 2022-04-27 | 2022-04-27 | FT19970003 | WH0006 |
| OD00000053 | P0005 | 6200 | 2022-04-15 | 2022-04-30 | 2022-05-01 | FT19940003 | WH0003 |
| OD00000054 | P0006 | 1000 | 2022-04-21 | 2022-04-30 | NULL | FT19940003 | WH0009 |
| OD00000055 | P0008 | 1500 | 2022-04-25 | 2022-05-11 | 2022-05-11 | FT19980003 | WH0009 |

#### 기대하는 출력 결과 (RESULT)
| ORDER_ID | PRODUCT_ID | OUT_DATE | 출고여부 |
| --- | --- | --- | --- |
| OD00000051 | P0002 | 2022-04-21 | 출고완료 |
| OD00000052 | P0003 | 2022-04-27 | 출고완료 |
| OD00000053 | P0005 | 2022-05-01 | 출고완료 |
| OD00000054 | P0006 | NULL | 출고미정 |
| OD00000055 | P0008 | 2022-05-11 | 출고대기 |


```sql
-- [문제 설명]
-- FOOD_ORDER 테이블에서 2022-05-01을 기준으로 출고여부를 분류합니다.
-- OUT_DATE <= 2022-05-01 이면 '출고완료', OUT_DATE > 2022-05-01 이면 '출고대기', OUT_DATE가 NULL이면 '출고미정'.
-- 출력 컬럼: ORDER_ID, PRODUCT_ID, OUT_DATE, 출고여부
-- 결과는 ORDER_ID 기준 오름차순 정렬합니다.
SELECT 
    ORDER_ID,
    PRODUCT_ID,
    DATE_FORMAT(OUT_DATE, '%Y-%m-%d') AS OUT_DATE,
    CASE 
        WHEN OUT_DATE <= '2022-05-01' THEN '출고완료'
        WHEN OUT_DATE > '2022-05-01' THEN '출고대기'
        ELSE '출고미정'
    END AS 출고여부
FROM FOOD_ORDER
ORDER BY ORDER_ID ASC;
```

## 문제 3: 멸종위기의 대장균 찾기 (Level 3)

- **문제 출처**: [프로그래머스 멸종위기의 대장균 찾기](https://school.programmers.co.kr/learn/courses/30/lessons/301651)

---

### 문제 설명

`ECOLI_DATA` 테이블은 실험실에서 배양한 대장균 개체들의 정보를 담은 테이블입니다.
`ID`, `PARENT_ID`, `SIZE_OF_COLONY`, `DIFFERENTIATION_DATE`, `GENOTYPE`는 각각 개체 ID, 부모 개체 ID, 개체군의 크기, 분화된 날짜, 개체의 형질을 나타내는 값을 의미합니다.

루트 대장균(최초의 대장균)은 `PARENT_ID` 값이 `NULL` 입니다.
루트 대장균의 세대(`GENERATION`)를 1로 하며, 자식 개체의 세대는 부모 개체의 세대보다 1 큽니다.

각 세대별로 자식이 없는 개체의 수(컬럼명: `COUNT`)와 세대(컬럼명: `GENERATION`)를 출력하는 SQL문을 작성해주세요.
이때 결과는 세대(`GENERATION`)에 대해 오름차순 정렬해주세요.

- 모든 세대에는 자식이 없는 개체가 최소 1개 이상 존재합니다.

---

### 예시 테이블 및 기대 결과

#### ECOLI_DATA (대장균 개체 정보)
| ID | PARENT_ID | SIZE_OF_COLONY | DIFFERENTIATION_DATE | GENOTYPE |
| --- | --- | --- | --- | --- |
| 1 | NULL | 10 | 2019-01-01 | 5 |
| 2 | NULL | 2 | 2019-01-01 | 3 |
| 3 | 2 | 100 | 2020-01-01 | 4 |
| 4 | 2 | 16 | 2020-01-01 | 4 |
| 5 | 2 | 17 | 2020-01-01 | 6 |
| 6 | 4 | 101 | 2021-01-01 | 22 |
| 7 | 4 | 101 | 2022-01-01 | 23 |
| 8 | 6 | 1 | 2022-01-01 | 27 |

#### 기대하는 출력 결과 (RESULT)
| COUNT | GENERATION |
| --- | --- |
| 1 | 1 |
| 2 | 2 |
| 1 | 3 |
| 1 | 4 |


```sql
-- [문제 설명]
-- ECOLI_DATA 테이블에서 루트 개체(PARENT_ID IS NULL)를 1세대로 하여 각 개체의 세대(GENERATION)를 구합니다.
-- 자식이 없는 개체(다른 개체의 PARENT_ID로 등장하지 않는 개체)를 세대별로 집계합니다.
-- 출력 컬럼: COUNT(자식 없는 개체 수), GENERATION
-- 결과는 GENERATION 기준 오름차순 정렬합니다.
WITH RECURSIVE ECOLI_GEN AS (
    SELECT ID, PARENT_ID, 1 AS GENERATION
    FROM ECOLI_DATA
    WHERE PARENT_ID IS NULL
    
    UNION ALL
    
    SELECT E.ID, E.PARENT_ID, G.GENERATION + 1 AS GENERATION
    FROM ECOLI_DATA E
    JOIN ECOLI_GEN G ON E.PARENT_ID = G.ID
)
SELECT 
    COUNT(*) AS COUNT,
    G.GENERATION
FROM ECOLI_GEN G
LEFT JOIN ECOLI_DATA C ON G.ID = C.PARENT_ID
WHERE C.ID IS NULL
GROUP BY G.GENERATION
ORDER BY G.GENERATION ASC;
```
