# 12일차 SQL 데이터 분석 학습

## 문제 1: 조건에 맞는 사용자와 총 거래금액 조회하기 (Level 3)

- **문제 출처**: [프로그래머스 조건에 맞는 사용자와 총 거래금액 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/164668)

---

### 문제 설명

`USED_GOODS_BOARD`와 `USED_GOODS_USER` 테이블에서 완료된 중고 거래의 총금액이 70만 원 이상인 사람의 회원 ID, 닉네임, 총거래금액을 조회하는 SQL문을 작성해주세요. 결과는 총거래금액을 기준으로 오름차순 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### USED_GOODS_BOARD (중고 거래 게시판 정보)
| BOARD_ID | WRITER_ID | TITLE | CONTENTS | PRICE | CREATED_DATE | STATUS | VIEWS |
| --- | --- | --- | --- | --- | --- | --- | --- |
| B0001 | cjh2971 | 아이폰 13 미니 | 상태 A급 | 750000 | 2022-03-20 | DONE | 120 |
| B0002 | dlghdud99 | 맥북 프로 | 2021년형 | 1200000 | 2022-03-21 | SALE | 230 |
| B0003 | cjh2971 | 에어팟 프로 | 미개봉 | 200000 | 2022-03-25 | DONE | 95 |
| B0004 | sangjo21 | 게이밍 모니터 | 27인치 | 150000 | 2022-03-27 | DONE | 40 |

#### USED_GOODS_USER (중고 거래 게시판 사용자 정보)
| USER_ID | NICKNAME | CITY | STREET_ADDRESS1 | STREET_ADDRESS2 | TLNO |
| --- | --- | --- | --- | --- | --- |
| cjh2971 | 제이 | 서울시 | 서초구 반포대로 | 101동 202호 | 01012345678 |
| dlghdud99 | 영구 | 서울시 | 강남구 테헤란로 | 303동 404호 | 01023456789 |
| sangjo21 | 상조 | 경기도 | 성남시 분당구 | 505동 606호 | 01034567890 |

#### 기대하는 출력 결과 (RESULT)
| USER_ID | NICKNAME | TOTAL_SALES |
| --- | --- | --- |
| cjh2971 | 제이 | 950000 |

#### 상세 작동 흐름
1. `STATUS = 'DONE'`인 완료 거래 데이터만 필터링합니다.
2. 작성자(`WRITER_ID`)별로 그룹화하여 거래 총금액(`SUM(PRICE)`)을 집계합니다.
3. 총금액이 700,000원 이상인 사용자만 `HAVING` 절을 통해 선별합니다. (`cjh2971`: 750,000 + 200,000 = 950,000원)
4. `USED_GOODS_USER` 테이블과 조인하여 닉네임을 가져온 뒤, `TOTAL_SALES` 오름차순으로 정렬합니다.

```sql
-- `USED_GOODS_BOARD`와 `USED_GOODS_USER` 테이블에서 완료된 중고 거래의 총금액이 70만 원 이상인 사람의 
--회원 ID, 닉네임, 총거래금액을 조회하는 SQL문을 작성해주세요. 결과는 총거래금액을 기준으로 오름차순 정렬해주세요.
SELECT B.USER_ID, B.NICKNAME, SUM(A.PRICE) AS TOTAL_SALES
FROM USED_GOODS_BOARD A JOIN USED_GOODS_USER B ON A.WRITER_ID = B.USER_ID
WHERE A.STATUS = 'DONE'
GROUP BY B.USER_ID,B.NICKNAME
HAVING SUM(A.PRICE) >= 700000
ORDER BY TOTAL_SALES ASC;
```

## 문제 2: 저자 별 카테고리 별 매출액 집계하기 (Level 4)

- **문제 출처**: [프로그래머스 저자 별 카테고리 별 매출액 집계하기](https://school.programmers.co.kr/learn/courses/30/lessons/144856)

---

### 문제 설명

`2022년 1월`의 도서 판매 데이터를 기준으로 저자 별, 카테고리 별 매출액(`TOTAL_SALES = 판매량 * 판매가`)을 구하여, 저자 ID(`AUTHOR_ID`), 저자명(`AUTHOR_NAME`), 카테고리(`CATEGORY`), 매출액(`TOTAL_SALES`) 리스트를 출력하는 SQL문을 작성해주세요.
결과는 저자 ID를 오름차순으로, 저자 ID가 같다면 카테고리를 내림차순으로 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### BOOK (도서 정보)
| BOOK_ID | CATEGORY | AUTHOR_ID | PRICE | PUBLISHED_DATE |
| --- | --- | --- | --- | --- |
| 1 | 인문 | 1 | 10000 | 2020-01-01 |
| 2 | 경제 | 1 | 9000 | 2021-02-05 |
| 3 | 인문 | 2 | 11000 | 2021-04-11 |

#### AUTHOR (저자 정보)
| AUTHOR_ID | AUTHOR_NAME |
| --- | --- |
| 1 | 홍길동 |
| 2 | 김영희 |

#### BOOK_SALES (도서 판매 정보)
| BOOK_ID | SALES_DATE | SALES |
| --- | --- | --- |
| 1 | 2022-01-10 | 2 |
| 1 | 2022-01-20 | 3 |
| 2 | 2022-01-15 | 4 |
| 3 | 2022-01-08 | 5 |

#### 기대하는 출력 결과 (RESULT)
| AUTHOR_ID | AUTHOR_NAME | CATEGORY | TOTAL_SALES |
| --- | --- | --- | --- |
| 1 | 홍길동 | 인문 | 50000 |
| 1 | 홍길동 | 경제 | 36000 |
| 2 | 김영희 | 인문 | 55000 |

```sql
-- 코드를 입력하세요
SELECT A.AUTHOR_ID,C.AUTHOR_NAME,A.CATEGORY,SUM(A.PRICE * B.SALES) AS TOTAL_SALES
FROM BOOK A JOIN BOOK_SALES B ON A.BOOK_ID = B.BOOK_ID JOIN AUTHOR C ON A.AUTHOR_ID = C.AUTHOR_ID
WHERE B.SALES_DATE LIKE '2022-01%'
GROUP BY A.AUTHOR_ID,C.AUTHOR_NAME,A.CATEGORY
ORDER BY A.AUTHOR_ID ASC, A.CATEGORY DESC;
```
