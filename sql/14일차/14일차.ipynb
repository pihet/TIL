# 14일차 SQL 데이터 분석 학습

## 문제 1: 즐겨찾기가 가장 많은 식당 정보 출력하기 (Level 3)

- **문제 출처**: [프로그래머스 즐겨찾기가 가장 많은 식당 정보 출력하기](https://school.programmers.co.kr/learn/courses/30/lessons/131123)

---

### 문제 설명

`REST_INFO` 테이블은 식당의 정보를 담은 테이블입니다.
`REST_ID`, `REST_NAME`, `FOOD_TYPE`, `VIEWS`, `FAVORITES`, `PARKING_LOT`, `ADDRESS`, `TEL`은 각각 식당 ID, 식당 이름, 음식 종류, 조회 수, 즐겨찾기 수, 주차장 유무, 주소, 전화번호를 의미합니다.

`REST_INFO` 테이블에서 음식종류별로 즐겨찾기수가 가장 많은 식당의 음식 종류, ID, 식당 이름, 즐겨찾기수를 조회하는 SQL문을 작성해주세요.
이때 결과는 음식 종류를 기준으로 내림차순 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### REST_INFO (식당 정보)
| REST_ID | REST_NAME | FOOD_TYPE | VIEWS | FAVORITES | PARKING_LOT | ADDRESS | TEL |
| --- | --- | --- | --- | --- | --- | --- | --- |
| 00001 | 은돼지식당 | 한식 | 1150345 | 734 | N | 서울특별시 중구 다산로 149 | 010-4484-8751 |
| 00002 | 하이가쯔네 | 일식 | 120034 | 112 | N | 서울시 중구 신당동 375-21 | NULL |
| 00003 | 따띠따띠뜨 | 양식 | 1234023 | 102 | N | 서울시 강남구 신사동 627-3 1F | 02-6397-1023 |
| 00004 | 스시사카우스 | 일식 | 1522074 | 230 | N | 서울시 강남구 신사동 627-27 | 010-9394-2554 |
| 00005 | 코슌스 | 일식 | 15301 | 123 | N | 서울특별시 강남구 언주로153길 | 010-1315-8729 |

#### 기대하는 출력 결과 (RESULT)
| FOOD_TYPE | REST_ID | REST_NAME | FAVORITES |
| --- | --- | --- | --- |
| 한식 | 00001 | 은돼지식당 | 734 |
| 일식 | 00004 | 스시사카우스 | 230 |

```sql
-- [문제 설명]
-- REST_INFO 테이블에서 음식종류(FOOD_TYPE)별로 즐겨찾기수(FAVORITES)가 가장 많은 식당을 조회합니다.
-- 출력 컬럼: 음식 종류, 식당 ID, 식당 이름, 즐겨찾기수 (FOOD_TYPE, REST_ID, REST_NAME, FAVORITES)
-- 결과는 음식 종류를 기준으로 내림차순 정렬합니다.
SELECT FOOD_TYPE,REST_ID,REST_NAME,FAVORITES
FROM REST_INFO 
WHERE (FOOD_TYPE,FAVORITES) IN (
    SELECT FOOD_TYPE, MAX(FAVORITES)
    FROM REST_INFO
    GROUP BY FOOD_TYPE
)
ORDER BY FOOD_TYPE DESC;
```

## 문제 2: 있었는데요 없었습니다 (Level 3)

- **문제 출처**: [프로그래머스 있었는데요 없었습니다](https://school.programmers.co.kr/learn/courses/30/lessons/59043)

---

### 문제 설명

동물 보호소에 들어온 모든 동물의 정보를 담은 `ANIMAL_INS` 테이블과, 입양이 진행된 동물의 정보를 담은 `ANIMAL_OUTS` 테이블이 있습니다.
`ANIMAL_OUTS` 테이블의 `ANIMAL_ID`는 `ANIMAL_INS`의 `ANIMAL_ID`의 외래 키입니다.

관리자의 실수로 일부 동물의 입양일이 잘못 입력되었습니다.
보호 시작일(`ANIMAL_INS.DATETIME`)보다 입양일(`ANIMAL_OUTS.DATETIME`)이 더 빠른 동물의 아이디와 이름을 조회하는 SQL문을 작성해주세요.
이때 결과는 보호 시작일이 빠른 순으로 조회해야 합니다.

---

### 예시 테이블 및 기대 결과

#### ANIMAL_INS (보호소 입소 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A350276 | Cat | 2017-08-13 13:50:00 | Normal | Jewel | Spayed Female |
| A381217 | Dog | 2017-07-08 09:41:00 | Sick | Cherokee | Neutered Male |

#### ANIMAL_OUTS (입양 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A350276 | Cat | 2018-01-28 17:51:00 | Jewel | Spayed Female |
| A381217 | Dog | 2017-06-09 18:51:00 | Cherokee | Neutered Male |

#### 기대하는 출력 결과 (RESULT)
| ANIMAL_ID | NAME |
| --- | --- |
| A381217 | Cherokee |


```sql
-- [문제 설명]
-- ANIMAL_INS(보호 시작일)와 ANIMAL_OUTS(입양일)를 ANIMAL_ID로 조인합니다.
-- 보호 시작일보다 입양일이 더 빠른 동물의 아이디와 이름(ANIMAL_ID, NAME)을 조회합니다.
-- 결과는 보호 시작일이 빠른 순서로 정렬합니다.
SELECT B.ANIMAL_ID, B.NAME
FROM ANIMAL_INS A JOIN ANIMAL_OUTS B ON A.ANIMAL_ID = B.ANIMAL_ID
WHERE A.ANIMAL_INS > B.ANIMAL_OUTS
ORDER BY A.DATETIME ASC;
```
