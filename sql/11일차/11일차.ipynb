# 11일차 SQL 알고리즘 학습

## 문제 1: 그룹별 조건에 맞는 식당 목록 출력하기 (Level 4)

- **문제 출처**: [프로그래머스 그룹별 조건에 맞는 식당 목록 출력하기](https://school.programmers.co.kr/learn/courses/30/lessons/131124)

---

### 문제 설명

`MEMBER_PROFILE`와 `REST_REVIEW` 테이블에서 리뷰를 가장 많이 작성한 회원의 리뷰들을 조회하는 SQL문을 작성해주세요. 회원 이름(`MEMBER_NAME`), 리뷰 텍스트(`REVIEW_TEXT`), 리뷰 작성일(`REVIEW_DATE`)을 출력하는 작성일 오름차순, 리뷰 작성일이 같다면 리뷰 텍스트 오름차순으로 정렬해주세요. (단, `REVIEW_DATE`의 데이트 포맷은 `YYYY-MM-DD` 형태여야 합니다.)

---

### 예시 테이블 및 기대 결과

#### MEMBER_PROFILE (회원 정보)
| MEMBER_ID | MEMBER_NAME | TLNO | GENDER | DATE_OF_BIRTH |
| --- | --- | --- | --- | --- |
| m1 | 김민지 | 01011112222 | W | 1992-05-15 |
| m2 | 이철수 | 01022223333 | M | 1988-11-20 |

#### REST_REVIEW (식당 리뷰 정보)
| REVIEW_ID | REST_ID | MEMBER_ID | REVIEW_SCORE | REVIEW_TEXT | REVIEW_DATE |
| --- | --- | --- | --- | --- | --- |
| r1 | rest1 | m1 | 5 | 맛있어요 | 2022-03-01 |
| r2 | rest2 | m1 | 4 | 친절해요 | 2022-03-05 |
| r3 | rest1 | m2 | 5 | 보통입니다 | 2022-03-02 |

#### 기대하는 출력 결과 (RESULT)
| MEMBER_NAME | REVIEW_TEXT | REVIEW_DATE |
| --- | --- | --- |
| 김민지 | 맛있어요 | 2022-03-01 |
| 김민지 | 친절해요 | 2022-03-05 |

#### 상세 작동 흐름
1. 회원별 리뷰 작성 개수를 집계하여 가장 리뷰를 많이 작성한 회원(m1 김민지, 총 2개)을 선별합니다.
2. 선별된 회원 m1이 작성한 모든 리뷰 데이터(r1, r2)를 가져옵니다.
3. 회원 이름, 리뷰 텍스트, 날짜 포맷(`YYYY-MM-DD`)을 적용한 리뷰 작성일을 추출합니다.
4. 리뷰 작성일 오름차순, 리뷰 텍스트 오름차순으로 정렬하여 출력합니다.

```sql
SELECT A.MEMBER_NAME, B.REVIEW_TEXT, DATE_FORMAT(B.REVIEW_DATE, '%Y-%m-%d') AS REVIEW_DATE
FROM MEMBER_PROFILE A LEFT JOIN REST_REVIEW B ON A.MEMBER_ID = B.MEMBER_ID
WHERE A.MEMBER_ID IN (
    SELECT MEMBER_ID
    FROM REST_REVIEW
    GROUP BY MEMBER_ID
    ORDER BY COUNT(*) DESC
    LIMIT 1 
    )
ORDER BY B.REVIEW_DATE ASC , B.REVIEW_TEXT ASC;
```

## 문제 2: 년, 월, 성별 별 상품 구매 회원 수 구하기 (Level 4)

- **문제 출처**: [프로그래머스 년, 월, 성별 별 상품 구매 회원 수 구하기](https://school.programmers.co.kr/learn/courses/30/lessons/131532)

---

### 문제 설명

`USER_INFO` 테이블과 `ONLINE_SALE` 테이블에서 년, 월, 성별 별로 상품을 구매한 회원수를 집계하는 SQL문을 작성해주세요. 결과는 년, 월, 성별을 기준으로 오름차순 정렬해주세요. 단, 성별 정보가 없는 경우 결과에서 제외해주세요.

---

### 예시 테이블 및 기대 결과

#### USER_INFO (회원 정보)
| USER_ID | GENDER | AGE | JOINED |
| --- | --- | --- | --- |
| 1 | 1 | 24 | 2021-06-01 |
| 2 | 0 | 29 | 2021-06-25 |
| 3 | NULL | 35 | 2021-07-01 |

#### ONLINE_SALE (상품 구매 정보)
| ONLINE_SALE_ID | USER_ID | PRODUCT_ID | SALES_AMOUNT | SALES_DATE |
| --- | --- | --- | --- | --- |
| 1 | 1 | 10 | 1 | 2022-01-15 |
| 2 | 1 | 20 | 2 | 2022-01-20 |
| 3 | 2 | 10 | 1 | 2022-01-18 |
| 4 | 3 | 30 | 1 | 2022-01-22 |

#### 기대하는 출력 결과 (RESULT)
| YEAR | MONTH | GENDER | USERS |
| --- | --- | --- | --- |
| 2022 | 1 | 0 | 1 |
| 2022 | 1 | 1 | 1 |

#### 상세 작동 흐름
1. 3번 유저는 `GENDER`가 `NULL`이므로 결과 집계 대상에서 제외됩니다.
2. 1번 유저(남성, GENDER=1)는 2022년 1월에 상품을 2번 구매했으나, 구매 회원 수는 고유 1명입니다.
3. 2번 유저(여성, GENDER=0)는 2022년 1월에 상품을 1번 구매하였으므로 구매 회원 수 1명입니다.
4. 년, 월, 성별 오름차순으로 그룹화하여 최종 회원 수를 반환합니다.

```sql
SELECT YEAR(B.SALES_DATE) AS YEAR, MONTH(B.SALES_DATE) AS MONTH , A.GENDER, COUNT(DISTINCT A.USER_ID) AS USERS
FROM USER_INFO A JOIN ONLINE_SALE B ON A.USER_ID = B.USER_ID
WHERE A.GENDER IS NOT NULL
GROUP BY YEAR(B.SALES_DATE), MONTH(B.SALES_DATE), A.GENDER
ORDER BY YEAR ASC, MONTH ASC, A.GENDER ASC;
```

## 문제 3: 조회수가 가장 높은 게시물에 대한 첨부파일 조회하기 (Level 3)

- **문제 출처**: [프로그래머스 조회수가 가장 높은 게시물에 대한 첨부파일 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/164671)

---

### 문제 설명

`USED_GOODS_BOARD`와 `USED_GOODS_FILE` 테이블에서 조회수가 가장 높은 중고거래 게시물에 대한 첨부파일 경로를 조회하는 SQL문을 작성해주세요. 첨부파일 경로는 `/get_result/BOARD_ID/FILE_IDFILE_NAMEFILE_EXT` 형식으로 출력해야 합니다. FILE ID를 기준으로 내림차순 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### USED_GOODS_BOARD (중고거래 게시판 정보)
| BOARD_ID | WRITER_ID | TITLE | PRICE | VIEWS | STATUS |
| --- | --- | --- | --- | --- |
| B0001 | user1 | 가방 팝니다 | 10000 | 150 | DONE |
| B0002 | user2 | 자전거 팝니다 | 50000 | 300 | SALE |

#### USED_GOODS_FILE (게시글 첨부파일 정보)
| FILE_ID | FILE_EXT | FILE_NAME | BOARD_ID |
| --- | --- | --- | --- |
| F0001 | .jpg | bike1 | B0002 |
| F0002 | .png | bike2 | B0002 |
| F0003 | .jpg | bag1 | B0001 |

#### 기대하는 출력 결과 (RESULT)
| FILE_PATH |
| --- |
| /get_result/B0002/F0002bike2.png |
| /get_result/B0002/F0001bike1.jpg |

#### 상세 작동 흐름
1. `USED_GOODS_BOARD` 테이블에서 `VIEWS` 값이 가장 큰 게시글을 찾습니다 (B0002, 조회수 300).
2. 해당 게시글(B0002)과 연관된 첨부파일 정보(F0001, F0002)를 `USED_GOODS_FILE`에서 조회합니다.
3. 지정된 포맷 `/get_result/BOARD_ID/FILE_IDFILE_NAMEFILE_EXT` 문자열로 결합합니다.
4. `FILE_ID`를 기준으로 내림차순 정렬하여 최종 출력을 완성합니다.

```sql
-- 코드를 작성해주세요.
-- 코드를 입력하세요
SELECT CONCAT('/home/grep/src/',B.BOARD_ID,'/',B.FILE_ID,B.FILE_NAME,B.FILE_EXT) AS FILE_PATH
FROM USED_GOODS_BOARD A JOIN USED_GOODS_FILE B ON A.BOARD_ID = B.BOARD_ID
WHERE A.VIEWS = (SELECT MAX(VIEWS) AS VIEWS
                FROM USED_GOODS_BOARD
                )
ORDER BY B.FILE_ID DESC;
```
